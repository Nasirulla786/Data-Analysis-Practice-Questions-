```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = r"C:\Users\ry356\Downloads\ChatGPT Image Oct 6, 2026, 10_20_59 AM.png"
c = cv2.imread(img)
plt.imshow(c)
plt.show()
```

```python
a = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(a)
plt.axis("on")
plt.show()
```

```python
a = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(a)
plt.axis("on")
plt.show()
```

```python
resized = cv2.resize(img,  (400,300))
print(resized.shape)
plt.imshow(resized)
```

```python
rotaed = cv2.rotate(img,cv2.ROTATE_90_CLOCKWISE)
plt.imshow(rotaed)
```

```python
rotaed = cv2.rotate(img,cv2.ROTATE_90_COUNTERCLOCKWISE)
plt.imshow(rotaed)
```

```python
cv2.line(img,(250,250),(500,500),(246,236,130),30)
plt.imshow(img)
```
