```python
import matplotlib.pyplot as plt

x = ["Mon","Tue","Wed","Thursdaya","Fri","Sat","Sun"]
y = [1000,2000,3000,1000,5000,6000,7000]
z = [3000,4000,5000,6000,7000,1000,2000]

plt.bar(x, y, label="Income")
plt.bar(x, z, label="Revenue" , width=0.5)

plt.xlabel("Days")
plt.ylabel("Income")
plt.title("Sales Chart")
plt.legend()
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [45, 50, 55, 60, 60, 65, 70, 75, 80, 85, 90]
plt.hist(x, bins=5 ,edgecolor="black")
plt.show()


# Internal Logic:---

# max = 90 
# min = 45 
# bins = 45/5 = 9

# 45+ 9 = 54
# 54 + 9 =63
# 63 + 9 = 72
# 72 + 9 = 81
# 81+9 = 90

# 45 - 54 - >  45 , 50 -> 2
# 54 - 63 - >  55 , 60 , 60 ->3
# 63 - 72 - >  65 , 70 ->2
# 72 - 81 - >  75 , 80 ->2
# 81 - 90 - >  85 , 90 ->2

# 2 3 2 2 2 
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "name": ["Nasir", "Rahul", "Aman", "Priya", "Sara", "Karan"],
    "age": [20, 21, 19, 20, 22, 21],
    "marks": [80, 90, 70, 95, 85, 75],
    "gender": ["M", "M", "M", "F", "F", "M"],
    "dept": ["BCA", "MCA", "BCA", "MCA", "BCA", "BCA"],
    "hours": [3, 5, 2, 6, 4, 3],
    "salary": [25000, 40000, 22000, 45000, 35000, 28000]
})

sns.lineplot(data = df, x = "name" , y = "marks" , hue="gender"  )
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
data = {
    "name": ["Nasir", "Rahul", "Aman", "Priya", "Sara", "Karan"],
    "age": [20, 21, 19, 20, 22, 21],
    "marks": [80, 90, 70, 95, 85, 75],
    "gender": ["M", "M", "M", "F", "F", "M"],
    "dept": ["BCA", "MCA", "BCA", "MCA", "BCA", "BCA"],
    "hours": [3, 5, 2, 6, 4, 3],
    "salary": [25000, 40000, 22000, 45000, 35000, 28000]
}
df = pd.DataFrame(data)
print(df)
v = sns.countplot(data=df , x = "age")
for i in v.containers:
    v.bar_label(i)
    
plt.show()
```
