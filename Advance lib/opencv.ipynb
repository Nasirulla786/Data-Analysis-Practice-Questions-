```python
import cv2
import matplotlib.pyplot as plt
img = cv2.imread("bg.png")
conCol = cv2.cvtColor(img , cv2.COLOR_BGR2RGB)
resImg = cv2.resize(conCol ,(300 ,500)) #width , height
cv2.line(resImg ,  (50 , 50) ,(200 , 100) ,(255,255,255) , 13)
cv2.putText(resImg , "Harshu" , (100 , 100) , cv2.FONT_HERSHEY_SIMPLEX , 2 , (255, 0 , 0 ) , 3)
plt.axis("off")
plt.imshow(resImg)



# print(img.shape)
# print(img.ndim)
# print(img.size)
# print(img.dtype)
```
