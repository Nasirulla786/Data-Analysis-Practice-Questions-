```python
# Boolean indexes

# check all elements that is greater or not 
# print all elements which is greater than 3 


k = np.array([1,2,3,4,5])
print(k)

# masking
# print(k[k>3])



print(k>3)
j = np.where(k>3,"out","no")
print(j)

z = np.where(k%2==0 ,"yes ","no")
print(z)
```
