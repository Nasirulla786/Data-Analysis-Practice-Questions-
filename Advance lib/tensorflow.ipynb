```python
import tensorflow as tf
# Scaler
v = tf.constant(10)
print(v)
print(v.shape)
print(v.dtype)
```

```python
# Vector
v = tf.constant([1,2,3,4,4])
print(v)
```

```python
# matrix
v = tf.constant([
    [1,2,3],
    [4,5,6],
    [7,8,9]
])
print(v)
```

```python
# matrix multipication
a = [
    [1,2], #row __
    [4,5],
]

b = [
    [4,5], #column |
    [7,8]
]


v = tf.matmul(a,b)
print(v)

# (1x4 + 2x7) , ( 1x5 + 2X8)
# (4x4 + 5x7 ),  (4x5 + 5x8)


# (4 + 14) , (5 + 16)
# (16 + 35) , (20 , 40)

# 18 , 21
# 51 , 60

```
