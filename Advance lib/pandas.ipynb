```python
# Data Cleanup
di = {
    "name":["nasirulla","harshu","raaz"],
    "age":[16,None,19],
    "roll":[None , 2 , 3]
}
df = pd.DataFrame(di)
print(df)


# df.isnull().sum()
df["age"] = df["age"].fillna(df["age"].mean())
df = df.fillna(0)
print(df)
```
